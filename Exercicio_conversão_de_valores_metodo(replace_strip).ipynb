```python
dados2 = pd.read_json('dados_locacao_imoveis.json')
```

```python
dados2 = pd.json_normalize(dados2['dados_locacao'])
```

```python
colunas = list(dados2.columns)
colunas
```

```python
dados2= dados2.explode(colunas[1:])
```

```python
dados2
```

```python
dados2['valor_aluguel'] = dados2['valor_aluguel'].apply(lambda x: x.replace('$', '').replace('reais', '').replace(',','.').strip()).astype(float)
```

```python
dados2['apartamento'] = dados2['apartamento'].str.replace('\(blocoAP\)', '', regex=True)
```

```python
dados2
```

```python
dados2.info()
```

```python
dados2['datas_de_pagamento'] = pd.to_datetime(dados2['datas_de_pagamento'], format='%d/%m/%Y')
```

```python
dados2['datas_combinadas_pagamento'] = pd.to_datetime(dados2['datas_combinadas_pagamento'], format='%d/%m/%Y')
```

```python
dados2['atraso'] = dados2['datas_combinadas_pagamento'] - dados2['datas_de_pagamento']
```

```python
media_atraso = dados2.groupby(['apartamento'])['atraso'].mean()
```
