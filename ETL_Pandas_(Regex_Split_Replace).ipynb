```python
import pandas as pd
```

```python
dados = pd.read_json('/content/dados_hospedagem.json')
```

```python
dados = pd.json_normalize(dados['info_moveis'])
```

```python
colunas = list(dados.columns)
colunas
```

```python
colunas_lista = []
colunas_nao_lista = []

for col in dados.columns:
    # Find the first non-null value in the column
    first_non_null = dados[col].dropna().iloc[0] if not dados[col].dropna().empty else None

    if isinstance(first_non_null, list):
        colunas_lista.append(col)
    else:
        colunas_nao_lista.append(col)

print("Colunas com valores de lista:", colunas_lista)
print("Colunas com valores não-lista:", colunas_nao_lista)
```

```python
dados = dados[colunas_nao_lista + colunas_lista]
print("DataFrame reordered successfully. New column order:")
print(dados.columns.tolist())
```

```python
colunas = list(dados.columns)
colunas
```

```python
dados_novos = dados.explode(colunas[3:])
dados_novos.head()
```

```python
dados_novos.reset_index(inplace=True, drop=True)
dados_novos.head()
```

```python
dados.info()
```

```python
import numpy as np
```

```python
dados_novos['max_hospedes'] = dados_novos['max_hospedes'].astype(np.int64)
```

```python
dados.info()
```

```python
col_numericas = ['quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']
```

```python
dados_novos[col_numericas] = dados_novos[col_numericas].astype(np.int64)
```

```python
dados_novos.info()
```

```python
dados_novos['avaliacao_geral'] = dados_novos['avaliacao_geral'].astype(np.float64)
```

```python
dados_novos.info()
```

```python
dados_novos['precos'] = dados_novos['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
dados_novos['precos'] = dados_novos['precos'].astype(np.float64)
```

```python
dados_novos.info()
```

```python
dados_novos[['taxa_deposito', 'taxa_limpeza']] = dados_novos[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
dados_novos[['taxa_deposito', 'taxa_limpeza']] = dados_novos[['taxa_deposito', 'taxa_limpeza']].astype(np.float64)
```

```python
dados_novos.info()
```

```python
dados_novos['descricao_local'] = dados_novos['descricao_local'].str.lower()
```

```python
dados_novos['descricao_local'][3169]
```

```python
dados_novos['descricao_local'] = dados_novos['descricao_local'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)
```

```python
dados_novos['descricao_local'] = dados_novos['descricao_local'].str.replace('(?<!\w)-(?!\w)',' ',regex=True)
```

```python
dados_novos['descricao_local'] = dados_novos['descricao_local'].str.split()
```

```python
dados_novos.head()
```

```python
dados_novos['comodidades'] = dados_novos['comodidades'].str.replace('\{|}|\"','',regex=True)
```

```python
dados_novos['comodidades'] = dados_novos['comodidades'].str.split(',')
```

```python
dados_novos.head()
```

```python
dados_novos['descricao_vizinhanca'] = dados_novos['descricao_vizinhanca'].str.replace('[^a-zA-Z0-9\-\']',' ',regex=True)
```
